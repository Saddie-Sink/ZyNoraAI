## Step 4: Train GNN Layout Model

```python
from training.models.gnn_layout_model import GNNLayoutModel, total_loss
from training.datasets.resplan_dataset import ResPlanDataset
from torch_geometric.loader import DataLoader
import torch.optim as optim
from torch.optim.lr_scheduler import CosineAnnealingLR
import time

DATA_DIR   = '/kaggle/working/data/resplan'
SAVE_DIR   = '/kaggle/working/models'
EPOCHS     = 100
BATCH_SIZE = 32
LR         = 1e-3
PATIENCE   = 12

train_ds = ResPlanDataset(DATA_DIR, split='train')
val_ds   = ResPlanDataset(DATA_DIR, split='val')
print(f'Train: {len(train_ds)} | Val: {len(val_ds)}')

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=2)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

model     = GNNLayoutModel(hidden_dim=64, gat_heads=4).to(device)
optimizer = optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-5)

best_val, patience_cnt = float('inf'), 0
train_losses, val_losses = [], []

for epoch in range(EPOCHS):
    model.train()
    epoch_loss, n = 0.0, 0
    t0 = time.time()
    for batch_data in train_loader:
        if batch_data is None: continue
        batch_data = batch_data.to(device)
        optimizer.zero_grad()
        pred = model(batch_data.x, batch_data.edge_index, batch_data.edge_attr,
                     batch_data.batch, batch_data.context.squeeze(1))
        loss, _ = total_loss(pred, batch_data.y, batch_data.x, batch_data.batch)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        epoch_loss += loss.item() * batch_data.num_graphs
        n += batch_data.num_graphs
    scheduler.step()

    # Validation
    model.eval()
    v_total, v_n = 0.0, 0
    with torch.no_grad():
        for bd in val_loader:
            if bd is None: continue
            bd = bd.to(device)
            p = model(bd.x, bd.edge_index, bd.edge_attr, bd.batch, bd.context.squeeze(1))
            l, _ = total_loss(p, bd.y, bd.x, bd.batch)
            v_total += l.item() * bd.num_graphs; v_n += bd.num_graphs
    train_loss = epoch_loss / max(1,n)
    val_loss   = v_total / max(1,v_n)
    train_losses.append(train_loss); val_losses.append(val_loss)
    marker = ' ← BEST' if val_loss < best_val else ''
    print(f'[{epoch+1:3d}/{EPOCHS}] train={train_loss:.4f} val={val_loss:.4f} ({time.time()-t0:.1f}s){marker}')

    if val_loss < best_val:
        best_val = val_loss; patience_cnt = 0
        torch.save({'epoch': epoch+1, 'model_state': model.state_dict(), 'best_val_loss': best_val},
                   f'{SAVE_DIR}/gnn_layout_best.pt')
    else:
        patience_cnt += 1
        if patience_cnt >= PATIENCE:
            print(f'Early stop at epoch {epoch+1}'); break

print(f'GNN Layout training done. Best val: {best_val:.4f}')
```

## Step 5: Download Structured3D + Train Extrusion Model

```python
# Structured3D is available on their official site (requires registration)
# OR use the synthetic fallback built into Structured3DDataset (no download needed)
# For real data: download from https://structured3d-dataset.org and upload to Kaggle as a dataset

from training.models.extrusion_model import ExtrusionModel, extrusion_loss
from training.datasets.structured3d_dataset import Structured3DDataset
from torch.utils.data import DataLoader as TorchDataLoader

# Using synthetic data if Structured3D not available
STRUCT3D_DIR = '/kaggle/input/structured3d/Structured3D'  # adjust if uploaded
if not __import__('os').path.exists(STRUCT3D_DIR):
    STRUCT3D_DIR = '/kaggle/working/data/structured3d_fake'
    __import__('os').makedirs(STRUCT3D_DIR, exist_ok=True)

def collate_fn(batch):
    xs = torch.stack([b[0] for b in batch])
    targets = {
        'wall_height_norm': torch.stack([b[1]['wall_height_norm'] for b in batch]),
        'floor_thick_norm': torch.stack([b[1]['floor_thick_norm'] for b in batch]),
        'ceiling_type':     torch.stack([b[1]['ceiling_type']     for b in batch]),
    }
    return xs, targets

ext_train = Structured3DDataset(STRUCT3D_DIR, split='train')
ext_val   = Structured3DDataset(STRUCT3D_DIR, split='val')
print(f'Extrusion train: {len(ext_train)} | val: {len(ext_val)}')

ext_train_loader = TorchDataLoader(ext_train, batch_size=128, shuffle=True,  collate_fn=collate_fn)
ext_val_loader   = TorchDataLoader(ext_val,   batch_size=128, shuffle=False, collate_fn=collate_fn)

ext_model = ExtrusionModel().to(device)
ext_opt   = optim.AdamW(ext_model.parameters(), lr=1e-3, weight_decay=1e-4)

best_ext, ext_patience = float('inf'), 0
EPOCHS_EXT = 80

for epoch in range(EPOCHS_EXT):
    ext_model.train()
    tl, tn = 0.0, 0
    for xs, tgts in ext_train_loader:
        xs = xs.to(device)
        tgts = {k: v.to(device) for k,v in tgts.items()}
        ext_opt.zero_grad()
        pred = ext_model(xs)
        loss, _ = extrusion_loss(pred, tgts)
        loss.backward()
        ext_opt.step()
        tl += loss.item()*xs.size(0); tn += xs.size(0)
    train_l = tl/max(1,tn)

    ext_model.eval()
    vl, vn = 0.0, 0
    with torch.no_grad():
        for xs, tgts in ext_val_loader:
            xs = xs.to(device); tgts = {k:v.to(device) for k,v in tgts.items()}
            pred = ext_model(xs); loss,_ = extrusion_loss(pred,tgts)
            vl+=loss.item()*xs.size(0); vn+=xs.size(0)
    val_l = vl/max(1,vn)
    marker = ' ← BEST' if val_l < best_ext else ''
    print(f'[Ext {epoch+1:3d}/{EPOCHS_EXT}] train={train_l:.4f} val={val_l:.4f}{marker}')
    if val_l < best_ext:
        best_ext = val_l; ext_patience = 0
        torch.save({'epoch':epoch+1,'model_state':ext_model.state_dict()},
                   f'{SAVE_DIR}/extrusion_model_best.pt')
    else:
        ext_patience += 1
        if ext_patience >= 10: print('Early stop'); break

print(f'Extrusion training done. Best val: {best_ext:.4f}')
```
